```python
data = open("Minor1.txt")

for i in data:
    if i.find(":") != -1:   # if the find method returns -1, it means that the substring was not found in the string. So, if it is not equal to -1, it means that the substring was found.
        speaker,line = i.split(":",1)   #max split ....control the split
        print(speaker,end=" ")
        print("said:",end=" ")
        print(line,end=" ")
    else:
        print(i)
```

# Try/except mechanism

try:
    your code which can cause error
except:
    your error recovery code

```python
#(Handling exception)

with open ("Minor1.txt") as data:
    
    for i in data:
        
        try:
            speaker,line = i.split(":")
            print(speaker,"said:",line)
        except:
            pass
```

```python
#(Handling exception)
try:
    with open ("Minor11.txt") as data:   # no file of this name is there, so it will give an error and the except block will be executed.
    
        for i in data:
        
            try:
                speaker,line = i.split(":")
                print(speaker,"said:",line)
            except:
                pass
except:
    print("File not found")
```

```python
try:
    with open ("Minor1.txt") as data: 
    
        for i in data:
        
            try:
                speaker,line = i.split(":")
                print(speaker,"said:",line)
            except:
                pass
except:
    print("File not found")
```

Be specific in your exceptions and don't go blindly for all errors.

```python
try:
    with open ("Minor11.txt") as data: 
    
        for i in data:
        
            try:
                speaker,line = i.split(":")
                print(speaker,"said:",line)
            except ValueError:                       #being specific about errors
                pass
except IOError:
    print("File not found")
```
