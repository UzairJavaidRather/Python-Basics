```python
data = open("Minor1.txt")

for i in data:
    if i.find(":") != -1:   # if the find method returns -1, it means that the substring was not found in the string. So, if it is not equal to -1, it means that the substring was found.
        speaker,line = i.split(":",1)   #max split ....control the split
        print(speaker,end=" ")
        print("said:",end=" ")
        print(line,end=" ")
    else:
        print(i)
```
