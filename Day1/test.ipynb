Python is a high-level, interpreted, dynamically typed, general-purpose programming language.

1. High-level :-> You write code using abstractions instead of directly manipulating CPU registers or memory addresses.

2. Interpreted :-> Source code is executed by a python interpreter(CPython compiles it to bytecode, which is then   executed by a virtual machine).

3. Dynamically typed :-> Variable types are determined at runtime.

4. Portable :-> windows, linux , macOS

5. Multi-paradigm :-> Supports procedural, object-oriented, and functional programming.
