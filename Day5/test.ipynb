# for Loops

```python
for i in "Python":
    print(i)
```

```python
for i in "Python":
    print("Welcome")   # This will print "Welcome" for each character in the string "Python"
    print(i,"Letter")  # This will print the character followed by the word "Letter"
```

```python
for i in range(5):
    print(i)
    print("Welcome")   # This will print "Welcome" for each number in the range 0 to 4
    print(i,i**i)   # This will print the number and its own power (i raised to the power of i) 
```

```python
# range(start, stop, step) is a built-in function in Python that generates a sequence of numbers.
# - start: The starting point of the sequence (inclusive). If not provided, it defaults to 0.
# - stop: The end point of the sequence (exclusive). This parameter is required.
# - step: The increment between each number in the sequence. If not provided, it defaults to 1.
# The range() function is commonly used in for loops to iterate over a sequence of numbers.

for i in range(1, 10, 2):
    print(i)  # This will print odd numbers from 1 to 9 (1, 3, 5, 7, 9)
```

```python
a = 0
b = 1

for i in range(5):
    c = a + b
    print(c)  # This will print the Fibonacci sequence starting from 1 (1, 2, 3, 5, 8)
    a = b
    b = c
```

# While Loop

```python
i = 0  # This initializes the variable i to 0, which will be used as a counter in the while loop.

while i < 5:  # This condition checks if i is less than 5. If it is, the loop will continue to execute.
    print(i)
    i = i + 1
```

```python
msg = ""

while msg != "exit":  # This condition checks if the variable msg is not equal to the string "exit". If it is not, the loop will continue to execute.
    msg = input("Enter a message (type 'exit' to quit): ")  # This prompts the user to enter a message and assigns it to the variable msg.
    if msg != "exit":
        print("You entered:", msg)  # This will print the message entered by the user, unless they typed "exit".
```

```python
i = 1

while i <= 10:
    if i % 2 == 0:  # This condition checks if i is even (i.e., divisible by 2 with no remainder).
        print(i)  # This will print the even number i.
    else:
        print(i)  # This will print the odd number i.
    i = i + 1  # This increments i by 1 in each iteration of the loop.
```

# Functions

```python
x = 10   #global variable, can be accessed anywhere in the code

def update():
    global x  # This tells Python that we want to use the global variable x instead of creating a new local variable.
    print(x)
    x = x + 10   # This updates the value of the global variable x to 20.
    print("Inside function:", x)

update()
print("Outside function:", x)  # This will print the updated value of x, which is now 20.
```

```python
ADA = 5

def os():
    ADA = 10
    def CN():
        nonlocal ADA
        ADA = ADA + 5  # nonlocal variable, can be accessed in the enclosing function os()
        print("Inside CN:", ADA)
        def DBMS():
            global ADA  # This tells Python that we want to use the global variable ADA instead of creating a new local variable.
            ADA = ADA + 20  # This updates the value of the global variable ADA
            print("Inside DBMS:", ADA)
        DBMS()
    CN()
os()
print("Outside function:", ADA)  # This will print the updated value of ADA, which should now be 25.
```
