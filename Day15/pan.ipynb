# Pandas

It is built on top of numpy and pandas.
It is widely used in data analysis, data science , machine learning.
Name comes from a term --- "Panel Data".
Here we work with series(1D labelled column) and Data Frame(2D data grid table).
With the help of this one can import,display,manipulate,export tabular data.

Hence , it is a python library used for data manipulation, analysis, and processing.

. Reading CSV or Excel files required a lot of code.
. Cleaning data(missing values,duplicates) had to be done manually.
. Filtering and searching data required writing long loops.
. Combining multiple datasets was complex.

# DataFrames

A tabular structure with rows and columns similar to Excel spreadsheets.

```python
import pandas as pd

data = {"Name":["Uzair","Mumtaz","Samin "],
        "Age":[23,20,22]}

df = pd.DataFrame(data)  #df === dataframe

print(df)
```

```python
import pandas as pd

data = {"Name":["Uzair","Mumtaz","Samin "],
        "Age":[23,20,22]}

df = pd.DataFrame(data, index=["i","ii","iii"])

print(df)
```

```python
import pandas as pd

data = {"Name":["Uzair","Mumtaz","Samin "],
        "Age":[23,20,22]}

df = pd.DataFrame(data, index=["i","ii","iii"])

print(df.loc["i"])
```

```python
import pandas as pd

data = {"Name":["Uzair","Mumtaz","Samin "],
        "Age":[23,20,22]}

df = pd.DataFrame(data, index=["i","ii","iii"])

print(df.iloc[1])
```

```python
import pandas as pd

data = {"Name":["Uzair","Mumtaz","Samin "],
        "Age":[23,20,22]}

df = pd.DataFrame(data, index=["i","ii","iii"])

#add a new column

df["Job"] = ["Cook","N/A","Teacher"]

#add a new row

new_row = pd.DataFrame([{"Name":"Uzaib","Age":21,"Job":"Engineer"}],index=["iv"])

#concatenate to existing one

df = pd.concat([df,new_row])

print(df)
```

```python
import pandas as pd

data = {"Name":["Uzair","Mumtaz","Samin "],
        "Age":[23,20,22]}

df = pd.DataFrame(data, index=["i","ii","iii"])

#add a new column

df["Job"] = ["Cook","N/A","Teacher"]

#add a new rows

new_row = pd.DataFrame([{"Name":"Uzaib","Age":21,"Job":"Engineer"},
                        {"Name":"Tehseen","Age":24,"Job":"Farmer"}],
                       index=["iv","v"]) #more dictionaries

#concatenate to existing one

df = pd.concat([df,new_row])

print(df)
```

# Filtering

Keeping the rows that match a condition

```python
tall_pok = df[df["Height"] >= 2]

print(tall_pok)   #gives data where height is greater or equal to 2
```

```python
heavy_pok = df[df["Weight"] > 100]

print(heavy_pok)
```

```python
legend = df[df["Legendary"] == 1]

print(legend)
```

# Aggregation

Reduce a set of values into a single value.

```python
print(df.mean(numeric_only=True))
```

```python
print(df.sum(numeric_only=True))
```

```python
print(df.min(numeric_only=True))
```

```python
print(df.max(numeric_only=True))
```

```python
print(df.count(numeric_only=True))
```

```python
# single column

# print(df["Height"].sum(numeric_only=True))

print(df["Height"].sum())  #min(),max(),count(),mean()

```

```python
group = df.groupby("Type1")

print(group["Height"].mean())
```

```python
group = df.groupby("Type1")

print(group["Height"].count())
```

```python
group = df.groupby("Type1")

print(group["Height"].max())
```

# Data Cleaning

It is the process of fixing/removing: incomplete, incorrect or irrelevant data.

75% of work done with Pandas is data cleaning.

```python
# 1. Drop irrelevant columns

df.drop(columns=["Legendary","No"])
```

```python
# 2. Handle missing data

df = df.dropna(subset=["Type2"])

print(df.to_string())
```

```python
df = df.fillna({"Type2":"None"})

print(df.to_string())
```

```python
# 3. Fix inconsistent values

df["Type1"] = df["Type1"].replace(({"Grass":"GRASS"}))

print(df.to_string())
```
