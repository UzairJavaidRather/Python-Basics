# Matplotlib

Matplotlib is the most popular Python library for creating graphs, charts, and visualizations.

We can create Line charts, Bar charts, Scatter plot, Histograms and Pie charts.

```python
import matplotlib.pyplot as plt

x = [2023,2024,2025,2026]
y = [15,25,30,20]

plt.plot(x,y) #plot() plotting of x and y

plt.show() #displays the plot
```

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.array([2023,2024,2025,2026])
y = np.array([15,25,30,20])

plt.plot(x,y) 

plt.show() 
```

# Plot Customization

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.array([2023,2024,2025,2026])
y1 = np.array([15,25,30,20])
y2 = np.array([17,23,38,5])


plt.plot(x,y1, marker = ".",  # Can use :-> " . , o , v , * "
         markersize = 15,       # markersize == ms
         markerfacecolor = "#1cd3fc",   # markerfacecolor == mfc
         markeredgecolor = "#1cd3fc",    # markeredgecolor == mec
         linestyle = "solid",               # or "dotted" or "dashdot" or "None"--for only points or "solid" or "dashed"
         linewidth = 4,
         color = "#1c5bfc")   # line color

plt.plot(x,y2, marker = ".",  # Can use :-> " . , o , v , * "
         markersize = 15,       # markersize == ms
         markerfacecolor = "#0b4e5d",   # markerfacecolor == mfc
         markeredgecolor = "#062a32",    # markeredgecolor == mec
         linestyle = "solid",               # or "dotted" or "dashdot" or "None"--for only points or "solid" or "dashed"
         linewidth = 4,
         color = "#1c5bfc")  

plt.show() 
```

```python
# Another way using dictionary

import matplotlib.pyplot as plt
import numpy as np

x = np.array([2023,2024,2025,2026])
y1 = np.array([15,25,30,20])
y2 = np.array([17,23,38,5])
y3 = np.array([15,28,33,27])

line_style = dict(marker = ".", 
                  markersize = 15,       
                  markerfacecolor = "#1cd3fc",   
                  markeredgecolor = "#1cd3fc",    
                  linestyle = "solid",              
                  linewidth = 4,
                  color = "#1c5bfc") 


plt.plot(x,y1,**line_style) # unpacking
plt.plot(x,y2 , **line_style)
plt.plot(x,y3 , **line_style)

plt.show()
```

# Labels

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.array([2023,2024,2025,2026])
y1 = np.array([15,25,30,20])
y2 = np.array([17,23,38,5])
y3 = np.array([15,28,33,27])

plt.title("Class Size",fontsize=25,
          family="Arial",
          fontweight="bold",
          color="#2d4cfc")  #gives title

#for X-axis

plt.xlabel("Year",fontsize=20,family="Arial",fontweight="bold",color="#2dbefc")

# for Y-axis

plt.ylabel("Students",fontsize=20,family="Arial",fontweight="bold",color="#2dbefc")

# for ticks

plt.tick_params(axis="both",colors="#2dbefc")   #both/x/y


plt.plot(x,y1) 
plt.plot(x,y2 )
plt.plot(x,y3)

plt.xticks(x)  #if we want text only at the required values

plt.show()
```

# Grid Lines

Helps make plots easier to read by adding reference lines.

```python
import matplotlib.pyplot as plt
import numpy as np

x = [1, 2, 3, 4, 5]
y = [5, 10, 15, 20, 25]

plt.grid(axis="y", # x can be used too or both
         linewidth=2,
         color="lightgray",
         linestyle="dashed")

plt.plot(x,y)

plt.show()
```

# Bar charts

Compares categories of data by representing each category with a bar.

```python
import matplotlib.pyplot as plt
import numpy as np

categories = np.array(["Grains","Fruit","Vegetables","Protein","Dairy","Sweets"])
values = np.array([4, 3, 2, 5, 3, 1])  #consumed

plt.bar(categories,values,color="skyblue")

plt.title("Daily Consumption")

plt.xlabel("Food")
plt.ylabel("Quantity")

plt.show()
```

```python
#Horizontal Bar chart

import matplotlib.pyplot as plt
import numpy as np

categories = np.array(["Grains","Fruit","Vegetables","Protein","Dairy","Sweets"])
values = np.array([4, 3, 2, 5, 3, 1])  #consumed

plt.barh(categories,values,color="skyblue")

plt.title("Daily Consumption")

plt.xlabel("Quantity")
plt.ylabel("Food")

plt.show()
```

# Pie Charts

Circular chart divided into slices to show percentages of the total.
Good for visualizing distributions among categories.

```python
import matplotlib.pyplot as plt
import numpy as np

categories = ["Freshmen","Sophomores","Juniors","Seniors"]
values = np.array([300, 250, 275, 225])
colors = ["red","yellow","blue","green"]

plt.pie(values,
        labels=categories,
        autopct="%1.1f%%",  #to add percentage
        colors = colors,
        explode=[0,0,0,0.1], #you want to explode
        shadow=True,
        startangle=90)  #rotate the pie chart

plt.title("Gcet")

plt.show()
```
