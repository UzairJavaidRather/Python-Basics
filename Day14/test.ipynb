# Plot Customization

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.array([2023,2024,2025,2026])
y1 = np.array([15,25,30,20])
y2 = np.array([17,23,38,5])


plt.plot(x,y1, marker = ".",  # Can use :-> " . , o , v , * "
         markersize = 15,       # markersize == ms
         markerfacecolor = "#1cd3fc",   # markerfacecolor == mfc
         markeredgecolor = "#1cd3fc",    # markeredgecolor == mec
         linestyle = "solid",               # or "dotted" or "dashdot" or "None"--for only points or "solid" or "dashed"
         linewidth = 4,
         color = "#1c5bfc")   # line color

plt.plot(x,y2, marker = ".",  # Can use :-> " . , o , v , * "
         markersize = 15,       # markersize == ms
         markerfacecolor = "#0b4e5d",   # markerfacecolor == mfc
         markeredgecolor = "#062a32",    # markeredgecolor == mec
         linestyle = "solid",               # or "dotted" or "dashdot" or "None"--for only points or "solid" or "dashed"
         linewidth = 4,
         color = "#1c5bfc")  

plt.show() 
```

```python
# Another way using dictionary

import matplotlib.pyplot as plt
import numpy as np

x = np.array([2023,2024,2025,2026])
y1 = np.array([15,25,30,20])
y2 = np.array([17,23,38,5])
y3 = np.array([15,28,33,27])

line_style = dict(marker = ".", 
                  markersize = 15,       
                  markerfacecolor = "#1cd3fc",   
                  markeredgecolor = "#1cd3fc",    
                  linestyle = "solid",              
                  linewidth = 4,
                  color = "#1c5bfc") 


plt.plot(x,y1,**line_style) # unpacking
plt.plot(x,y2 , **line_style)
plt.plot(x,y3 , **line_style)

plt.show()
```

# Labels

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.array([2023,2024,2025,2026])
y1 = np.array([15,25,30,20])
y2 = np.array([17,23,38,5])
y3 = np.array([15,28,33,27])

plt.title("Class Size",fontsize=25,
          family="Arial",
          fontweight="bold",
          color="#2d4cfc")  #gives title

#for X-axis

plt.xlabel("Year",fontsize=20,family="Arial",fontweight="bold",color="#2dbefc")

# for Y-axis

plt.ylabel("Students",fontsize=20,family="Arial",fontweight="bold",color="#2dbefc")

# for ticks

plt.tick_params(axis="both",colors="#2dbefc")   #both/x/y


plt.plot(x,y1) 
plt.plot(x,y2 )
plt.plot(x,y3)

plt.xticks(x)  #if we want text only at the required values

plt.show()
```

# Grid Lines

Helps make plots easier to read by adding reference lines.

```python
import matplotlib.pyplot as plt
import numpy as np

x = [1, 2, 3, 4, 5]
y = [5, 10, 15, 20, 25]

plt.grid(axis="y", # x can be used too or both
         linewidth=2,
         color="lightgray",
         linestyle="dashed")

plt.plot(x,y)

plt.show()
```

# Bar charts

Compares categories of data by representing each category with a bar.

```python
import matplotlib.pyplot as plt
import numpy as np

categories = np.array(["Grains","Fruit","Vegetables","Protein","Dairy","Sweets"])
values = np.array([4, 3, 2, 5, 3, 1])  #consumed

plt.bar(categories,values,color="skyblue")

plt.title("Daily Consumption")

plt.xlabel("Food")
plt.ylabel("Quantity")

plt.show()
```

```python
#Horizontal Bar chart

import matplotlib.pyplot as plt
import numpy as np

categories = np.array(["Grains","Fruit","Vegetables","Protein","Dairy","Sweets"])
values = np.array([4, 3, 2, 5, 3, 1])  #consumed

plt.barh(categories,values,color="skyblue")

plt.title("Daily Consumption")

plt.xlabel("Quantity")
plt.ylabel("Food")

plt.show()
```

# Pie Charts

Circular chart divided into slices to show percentages of the total.
Good for visualizing distributions among categories.

```python
import matplotlib.pyplot as plt
import numpy as np

categories = ["Freshmen","Sophomores","Juniors","Seniors"]
values = np.array([300, 250, 275, 225])
colors = ["red","yellow","blue","green"]

plt.pie(values,
        labels=categories,
        autopct="%1.1f%%",  #to add percentage
        colors = colors,
        explode=[0,0,0,0.1], #you want to explode
        shadow=True,
        startangle=90)  #rotate the pie chart

plt.title("Gcet")

plt.show()
```

# Scatter Graph

Shows the relationship between two variables.
Helps to identify a correlation (+,-,None)
Example: study hours vs Test scores

```python
import matplotlib.pyplot as plt
import numpy as np

x1 = np.array([0, 1, 1, 2, 3, 4, 5, 6, 7, 7, 8]) #study hours
y1 = np.array([55, 60, 65, 62, 68, 70, 75, 78, 82, 85, 87]) #values of test scores (%)

x2 = np.array([0, 1, 2, 2, 3, 4, 5, 6, 7, 8, 8])
y2 = np.array([50, 58, 65, 70, 72, 78, 83, 88, 92, 95, 97]) 


plt.scatter(x1,y1, color="skyblue",alpha=0.5, #transparency values
            s=200,
            label="Class A")

plt.scatter(x2,y2, color="red",alpha=0.5,
            s=200,
            label="Class B")

plt.xlabel("Hours Studied")
plt.ylabel("Grade")

plt.title("Test Scores")

plt.legend() #gives details like classes in the graph
plt.show()
```

# Histograms
 A visual representation of the distribution of quantitative data.
 They group values into bins(intervals) and counts how many fall in each range.

```python
import matplotlib.pyplot as plt
import numpy as np

scores = np.random.normal(loc=80,scale=10, size=100) #scores around the 80 will be frequent and standard deviation of 10 (smaller no == more numbers will clutter around centre i.e 80) and size of 100.

scores = np.clip(scores,0,100) #clip scores below "0" and above "100"

plt.hist(scores,bins=8, #no. of bars
         color="lightgreen",
         edgecolor="black")

plt.title("Exam Scores")
plt.xlabel("Score")
plt.ylabel("Number of students")

plt.show()
```

# Subplots

. Figure = The entire Canvas
. Ax = A single plot (subplot)

```python
import matplotlib.pyplot as plt
import numpy as np

print(plt.subplots(2,2)) #return a tuple
```

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.array([1, 2, 3, 4, 5])

figure, axes = plt.subplots(2,2) # 4 subplots   ---(row,column)

axes[0,0].plot(x,x*2,color="red") #row=0 and column=0
axes[0,0].set_title("X*2")

axes[0,1].plot(x,x*4,color="green")
axes[0,1].set_title("X*4")

axes[1,0].plot(x,x*6,color="blue")
axes[1,0].set_title("X*6")

axes[1,1].plot(x,x*8,color="yellow")
axes[1,1].set_title("X*8")

plt.tight_layout() #prevents the overlapping of labels
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.array([1, 2, 3, 4, 5])

figure, axes = plt.subplots(2,2) 

axes[0,0].plot(x,x*2,color="red")
axes[0,0].set_title("X*2")

axes[0,1].bar(x,x**2,color="green")
axes[0,1].set_title("X**2")

axes[1,0].plot(x,x*6,color="blue")
axes[1,0].set_title("X*6")

axes[1,1].plot(x,x**4,color="yellow")
axes[1,1].set_title("X**4")

plt.tight_layout() #prevents the overlapping of labels
plt.show()
```
