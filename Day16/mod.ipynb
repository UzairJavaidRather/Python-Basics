# Module

A module is a file that contains python code(functions,classes,variables, or runnable statements) that you can resuse in other python programs.
A module usually has a .py extension.

# Math module

Mainly used for calculations involving numbers, geometry, trignometry, logrithms, and scientific computations.

```python
import math

print(math.sqrt(25))
```

```python
print(math.pow(2,3))
```

```python
print(math.factorial(5))
```

```python
print(math.ceil(4.3))
print(math.floor(4.8))
```

```python
print(math.sin(math.pi/2))
print(math.cos(0))
```

```python
print(math.pi)
print(math.e)  #eulers number
```

```python
radius = 6.6

area = math.pi * math.pow(radius,2)

print(area)
```

```python
print(math.log(10))
```

```python
print(math.log(4,2)) # log with specific base
```

```python
print(math.log2(8))
```
