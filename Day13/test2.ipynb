It is a programming paradigm that organizes code into objects and classes.
It helps in code reusability, security, and easy maintenance.

```python
# Class : It is a blueprint for creating objects.

class student:
    pass
```

# Constructor

A special method in a class that is automatically called when an object is created.

__init__(self,fullname)   ----here self,fullname are the arguments or attributes.

self : It is the reference to the curent object of the class, and is used to acess variables that belongs to the class.

```python
class student:
    
    #default constructor
    def __init__(self):
        print("Default Constructor is called") #it will not print "Default Constructor is called" because we are using parameterized constructor below.
        self.name = "Bob" # instance variable
        
    #parameterized constructor
    def __init__(self, fname, fage):
        print("Parameterized Constructor is called")
        self.name = fname
        self.age = fage
        
s1 = student("Uzair", 20) # s1 is an object of class student
print(s1.name) # Uzair
print(s1.age) # 20
print(s1.name, s1.age) # Uzair 20
```

In python ,a class can have only one constructor. Python does not support constructor overloading directly like in C++ or Java.
In above example second __init__() overrides the first one.Only the parametrized constructor exist.

# Class and Instance attributes

```python
#1. Class Attributes : A class attribute is shared among all instances of the class.

class student:
    school = "ABC School" # class attribute
    
s1 = student() # s1 is an object of class student
s2 = student() # s2 is an object of class student

print(s1.school) # ABC School
print(s2.school) # ABC School
```

```python
#2. Instance Attributes : Instance attributes belong to individual objects

class student:
    def __init__(self, fname, fage):
        self.name = fname # instance variable
        self.age = fage # instance variable
        
s1 = student("Uzair", 20) # s1 is an object of class student
s2 = student("Bob", 25) # s2 is an object of class student

print(s1.name, s1.age) # Uzair 20
print(s2.name, s2.age) # Bob 25
```

```python
# 3. Using Both Together

class student:
    school = "ABC School" # class attribute
    name = "any name" # class attribute   --- but it will be overridden by instance variable name in __init__ method as instance variable has more priority than class variable known as shadowing.
    
    def __init__(self, fname, fage):
        self.name = fname # instance variable
        self.age = fage # instance variable
        
s1 = student("Uzair", 20) # s1 is an object of class student
s2 = student("Bob", 25) # s2 is an object of class student

print(s1.name, s1.age, s1.school) # Uzair 20 ABC School
print(s2.name, s2.age, s2.school) # Bob 25 ABC School
```

```python
# 4. Modifying Class Attributes

class student:
    school = "ABC School" # class attribute
    
s1 = student() # s1 is an object of class student
s2 = student() # s2 is an object of class student

print(s1.school) # ABC School
print(s2.school) # ABC School

student.school = "XYZ School" # modifying class attribute

print(s1.school) # XYZ School
print(s2.school) # XYZ School

```

# Methods in OOPS

A method is a function defined inside a class that describes the behavior of objects.

```python
class student:
    def __init__(self, name, age):
        self.name = name
        self.age = age
    
    def display(self):
        print("Name:", self.name)
        
    def age_display(self):
        return self.age
        
s1 = student("Uzair", 20) # s1 is an object of class student
s1.display() # Name: Uzair
print(s1.age_display()) # 20
```

Q. Create a student that takes name and marks of 3 subjects as arguments in constructor.Then create a method to print the average.

```python
class student:
    def __init__(self,fname,dsa,coa,python):
        self.name = fname
        self.dsa = dsa
        self.coa = coa
        self.python = python
        
    def display(self):
        print("Name:", self.name)
        print("DSA Marks:", self.dsa)
        print("COA Marks:", self.coa)
        print("Python Marks:", self.python)
        print("Average Marks:", (self.dsa + self.coa + self.python)/3)
        
s1 = student("Uzair", 85, 90, 95) # s1 is an object of class student
s1.display()
```

```python
class Employee:
    company = "Google"

    def __init__(self, name):
        self.name = name

    def show_name(self):            # Instance method
        print(self.name)

    @classmethod              #decorator for class method
    def show_company(cls):          # Class method  : works with class attributes.Uses "cls" keyword to access class attributes. 
        print(cls.company)

    @staticmethod
    def greet():                    # Static method : does not access class or instance attributes. It is a utility function. Don't use "self" or "cls" keyword in static method.
        print("Welcome!")

e1 = Employee("Uzair")

e1.show_name()
Employee.show_company()
Employee.greet()
```

# Decorators :
They allow us to wrap another function in order to extend the behaviour of the wrapped function ,without permanently modifying it.

```python
class student:
    def __init__(self,fname,fmarks):
        self.name = fname
        self.marks = fmarks
        
    def display(self):
        sum = 0
        for mark in self.marks:
            sum += mark
            
        print("Average Marks:", sum/len(self.marks))
        
s1 = student("Uzair",[ 85, 90, 95]) # s1 is an object of class student
s1.display()
```

# Abstraction
Abstraction means hiding unnecessary implementation details and showing only the essential features to the user.

Real-Life Example
When you drive a car:
You use the steering wheel, brakes, and accelerator.
You do not need to know how the engine works internally.

The complex implementation is hidden from you.

Important Points
. Abstract classes are created using ABC.
. Abstract methods use @abstractmethod.
. An abstract class cannot be instantiated directly.

```python
class car:
    def __init__(self):
        self.acc = False
        self.speed = False
    
    def start(self):
        self.acc = True        # here we do not print unnessary details only print the required details which is "Car started" which is known as abstraction.
        self.speed = True
        print("Car started.")
        
c1 = car()
c1.start()
```

# Encapsulation :

Encapsulation means wrapping data and methods into a single unit (class) and controlling access to the data.

Real-Life Example
Consider a bank account:
Balance should not be changed directly.
Access is provided through methods such as deposit and withdraw.

```python
class Account:
    def __init__(self, account_number, balance):
        self.account_number = account_number
        self.balance = balance

    def credit(self, amount):
        self.balance += amount
        print(f"₹{amount} credited successfully.")

    def debit(self, amount):
        if amount <= self.balance:
            self.balance -= amount
            print(f"₹{amount} debited successfully.")
        else:
            print("Insufficient balance.")

    def print_balance(self):
        print(f"Account Number: {self.account_number}")
        print(f"Current Balance: ₹{self.balance}")


# Creating an account object
acc1 = Account("ACC1001", 5000)

# Credit amount
acc1.credit(2000)

# Debit amount
acc1.debit(1500)

# Print balance
acc1.print_balance()
```

# del Keyword

The del keyword is used to delete objects, variables, list elements, dictionary items, or object attributes.

```python
class student:
    def __init__(self, name, age):
        self.name = name
        self.age = age

s1 = student("Uzair", 20)
print(s1.name) # Uzair
print(s1.age) # 20

del s1.name # deleting instance variable name
print(s1.name) # AttributeError: 'student' object has no attribute 'name'

del s1 # deleting object s1
print(s1) # NameError: name 's1' is not defined
```

```python
# Python provides a special destructor method called __del__().

class Student:
    def __del__(self):
        print("Object destroyed")

s1 = Student()

del s1
```

# Private Class attributes and methods :

A private class attribute is a class attribute that is intended to be hidden from direct access outside the class.

In Python, private attributes are created by prefixing the attribute name with double underscores (__).

Same for method too.

```python
class Person:
    
    __name = "Uzair" # private class variable
    
    def __hello(self):
        print("Hello") # private method
        
    def display(self):
        self.__hello() # accessing private method inside the class
        print("Name:", self.__name) # accessing private class variable inside the class
        
p1 = Person()
p1.display() # Hello Name: Uzair
    
```

# Inheritance :

Inheritance is an OOP concept where one class acquires the properties (attributes) and behaviors (methods) of another class.

The existing class is called the Parent Class (Base Class/Super Class).
The new class is called the Child Class (Derived Class/Sub Class).

```python
# 1. Single Inheritance : A class can inherit from a single parent class.

class Parent:
    def show(self):
        print("Parent class")

class Child(Parent):
    pass

c = Child()
c.show() # Parent class
```

```python
class car:
    
    @staticmethod
    def start():
        print("Car started.")
        
class brand(car):                   # child class brand inherits from parent class car
    def __init__(self,fname):
        self.name = fname
        
b1 = brand("BMW")
b1.start() # Car started.

print(b1.name) # BMW
```

```python
# 2. Multilevel Inheritance : A class can inherit from a parent class, which in turn inherits from another parent class.

class Grandparent:
    def gp(self):
        print("Grandparent")

class Parent(Grandparent):
    def p(self):
        print("Parent")

class Child(Parent):
    def c(self):
        print("Child")

obj = Child()

obj.gp()
obj.p()
obj.c()
```

```python
# 3. Multiple Inheritance : A class can inherit from multiple parent classes.

class Father:
    def father_method(self):
        print("Father")

class Mother:
    def mother_method(self):
        print("Mother")

class Child(Father, Mother):
    def child_method(self):
        print("Child")

c = Child()

c.father_method()
c.mother_method()
c.child_method()
```

```python
# 4. Hierarchical Inheritance : Multiple classes can inherit from a single parent class.

class Animal:
    def eat(self):
        print("Eating")

class Dog(Animal):
    def bark(self):
        print("Barking")

class Cat(Animal):
    def meow(self):
        print("Meowing")

d = Dog()
c = Cat()

d.eat()
d.bark()
c.eat()
c.meow()
c.eat()
```

```python
# super() function : It is used to call the parent class method and constructor from the child class.

class car:
    def __init__(self,type):
        self.type = type
    
    @staticmethod
    def start():
        print("Car started.")
        
class brand(car):
    def __init__(self,fname,type):
        super().__init__(type) # calling parent class constructor
        super().start() # calling parent class method
        self.name = fname

b1 = brand("BMW", "SUV")
print(b1.name) # BMW
print(b1.type) # SUV
```

```python
#Class methods

class person:
    name = "anyone" # class variable
    
    @classmethod              #decorator for class method
    def change_name(cls,name):          # Class method  : works with class attributes.Uses "cls" keyword to access class attributes. 
        cls.name = name # modifying class variable name using class method
        
p1 = person()
print(p1.name) # anyone
p1.change_name("Uzair") # changing class variable name using class method
print(p1.name) # Uzair
        
```

```python
# property decorator : It is used to create getter and setter methods for class attributes.

class student:
    def __init__(self,fname,dsa,coa,python):
        self.name = fname
        self.dsa = dsa
        self.coa = coa
        self.python = python

    @property
    def average(self):
        sum = self.dsa + self.coa + self.python
        return sum / 3

s1 = student("Uzair",85, 90, 95) # s1 is an object of class student
print(s1.average) # 90.0

s1.dsa = 80 # modifying dsa marks

print(s1.average) # 88.33333333333333
```
