1. nd_array can be operated in vectorized form by taking benefit from SIMD stream.
2. No overhead of typechecking.
3. Easy to load and save files.
4. Other packages can efficiently interact with Numpy.
5. Provides inbuilt implementation for alegbra,statistics,broadcasting,etc.


Pros => Native speed , Homogenity, Contigious memory.

```python
%%time

N = 100000000

l1 = list(range(N)) #  it will create a list from 0 to N-1

for i in range(N):
    l1[i] = l1[i] * l1[i] # it will square each element in the list
```

```python
%%time

N = 100000000

l2 = list(range(N)) #  it will create a list from 0 to N-1

l2 = [i * i for i in l2] # it will square each element in the list using list comprehension i.e inline code
```

```python
%%time

N = 100000000

l3 = list(range(N)) #  it will create a list from 0 to N-1

# map(function, iterable) --> it will apply the function to each element in the iterable and return a map object which is an iterator

l3 = map(lambda x: x * x, l3) # it will square each element in the list using map function and lambda function i.e anonymous function
```
